```python
import os
from groq import Groq

groq_api_key = "groqapi"
client = Groq(api_key=groq_api_key)

def chat_groq(prompt, model="llama-3.1-8b-instant"):
    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}]
    )
    return response.choices[0].message.content

# Test
print(chat_groq("What is LLaMA3? Answer in 3 sentences."))
```

```python
# Multi-turn conversation via Groq
history = []

def chat_with_history(user_msg):
    history.append({"role": "user", "content": user_msg})
    response = client.chat.completions.create(
        model="llama-3.1-8b-instant",
        messages=history
    )
    reply = response.choices[0].message.content
    history.append({"role": "assistant", "content": reply})
    return reply

print(chat_with_history("My name is Alex. I study at RVU."))
print(chat_with_history("What is my name and where do I study?"))
```
